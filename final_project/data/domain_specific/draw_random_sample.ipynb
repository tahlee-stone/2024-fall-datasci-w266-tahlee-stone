# Random sample for human annotations

```python
import pandas as pd
import random
from pathlib import Path
import os
import json
import sys

# Load the dataset
df = pd.read_csv("data/domain_specific/banking_77_1k_triplet_set_long.csv") 

# Sample 25 PASS and 25 FAIL examples
random.seed(42)
sample_pass = df[df['label'] == 'PASS'].sample(n=25, random_state=42)
sample_fail = df[df['label'] == 'FAIL'].sample(n=25, random_state=42)

# Combine and shuffle the sample
sample_df = pd.concat([sample_pass, sample_fail]).sample(frac=1, random_state=42).reset_index(drop=True)

# Save the full sample with labels
sample_df.to_csv("data/domain_specific/sample_50_human_annotation.csv", index=False)
```
